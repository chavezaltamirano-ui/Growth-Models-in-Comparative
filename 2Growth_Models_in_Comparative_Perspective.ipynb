```python
import pandas as pd

# 1. Cargar panel base y GDD
panel = pd.read_csv("data_clean/china_us_panel_1990_2025.csv")
gdd   = pd.read_csv("data_raw/imf_gdd_debt_cn_us_1990_2024.csv")

print("Panel columnas:", panel.columns)
print("GDD columnas:", gdd.columns)

# 2. Asegurar tipos de 'year' (por si acaso)
panel["year"] = panel["year"].astype(int)
gdd["year"]   = gdd["year"].astype(int)

# 3. Hacer el merge por country-year
panel_gdd = panel.merge(
    gdd,
    on=["country", "year"],
    how="left"
)

print("\nVista rápida panel + GDD (head):")
print(panel_gdd.head())

print("\nVista rápida panel + GDD (tail):")
print(panel_gdd.tail())

print("\nNúmero de columnas:", panel_gdd.shape[1])

# 4. Guardar el panel extendido con deuda IMF GDD
out_path = "data_clean/china_us_panel_1990_2025_with_gdd_debt.csv"
panel_gdd.to_csv(out_path, index=False)
print("\nGuardado:", out_path)
```

```python
import pandas as pd

# Panel con PWT + WDI macro/fin + BIS + productividad + PPP
panel_full = pd.read_csv(
    "data_clean/china_us_panel_1990_2025_with_wdi_prod_gdpppp.csv"
)

# Deuda IMF GDD (1990–2024)
gdd = pd.read_csv("data_raw/imf_gdd_debt_cn_us_1990_2024.csv")

# Deuda WEO (2017–2026)
weo = pd.read_csv("data_raw/weo_gov_debt_cn_us_2017_2026.csv")

# Asegurar tipos y claves
panel_full["year"] = panel_full["year"].astype(int)
gdd["year"] = gdd["year"].astype(int)
weo["year"] = weo["year"].astype(int)

print("Panel_full columnas:", panel_full.columns)
print("GDD columnas:", gdd.columns)
print("WEO columnas:", weo.columns)

# Supongo que WEO tiene columnas: country, year, weo_gov_debt_gdp (ajusta al nombre real)
# Primero fusionamos IMF GDD:
super_panel = panel_full.merge(
    gdd,
    on=["country", "year"],
    how="left"
)

# Luego añadimos WEO (ajustando al nombre real de su columna de deuda)
# Por ejemplo si la columna es 'weo_pub_debt_gdp':
super_panel = super_panel.merge(
    weo,
    on=["country", "year"],
    how="left"
)

print("\nVista rápida super panel (head):")
print(super_panel.head())
print("\nVista rápida super panel (tail):")
print(super_panel.tail())
print("\nNúmero de columnas:", super_panel.shape[1])

out_path = "data_clean/china_us_super_panel_1990_2025.csv"
super_panel.to_csv(out_path, index=False)
print("\nGuardado:", out_path)
```
